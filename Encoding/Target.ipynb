```python
import pandas as pd
from sklearn.model_selection import train_test_split
from category_encoders import TargetEncoder
from sklearn.preprocessing import OneHotEncoder,LabelEncoder
```

```python
data = {
    # "Dept":["ML","DL","DL","Ml","DL","ML","DL","DL","Ml","DL"],
    "City" : ["Banglore","Vellore","Chennai","Banglore","Madurai","Banglore","Vellore","Chennai","Banglore","Madurai"],
    "Product" : ["Mobile","Clothing","Sandals","Electronics","Clothing","Sandals","Mobile","Clothing","Appliances","Furnichers"],
    "Price" : [2000,1000,4000,3000,1000,2000,1000,4000,3000,1000]
}
```

```python
df = pd.DataFrame(data)
```

# Scenario 1

**Assigning X and Y**

```python
x = df.drop(columns='Price')
y = df['Price']
```

```python
xtrain,xtest,ytrain,ytest = train_test_split(x,y,train_size=0.8,random_state=42)
```

```python
encoder = TargetEncoder(handle_unknown='value',cols=['City','Product'])
```

```python
encoder.fit_transform(xtrain,ytrain)
```

```python
encoder.transform(xtest)
```
