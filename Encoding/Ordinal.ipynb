```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OrdinalEncoder
```

```python
data = {
    "name":["A","C","D","B","E","A","E","C"],
    "dept_id":["s1","s4","s5","s2","s3","s1","s4","s5"],
    "d_name" :["Analyst","Developer","Engineer","Trainer","UI Developer","Developer","Engineer","Analyst"],
    "salary" :[15000,20000,23000,18000,21000,19000,25000,27000]
}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
x = df.drop(columns='salary')
y = df['salary']
```

```python
xtrain,xtest,ytrain,ytest = train_test_split(x,y,train_size=0.8,random_state=42)
```

# Scenario 1

* Error --> ValueError: Shape mismatch: if categories is an array, it has to be of shape (n_features,)


```python
encoder = OrdinalEncoder(categories=[['Developer','Analyst','Enginneer','Trainer',"UI Developer"],["s1",'s2','s5','s4','s3']],
                         handle_unknown="use_encoded_value",
                         unknown_value=-1
                         )

encoder.fit_transform(xtrain[['d_name','dept_id']])
# encoder.fit_transform(xtrain[['d_name','dept_id',"name"]])

#Error --> ValueError: Shape mismatch: if categories is an array, it has to be of shape (n_features,)
```
