```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder 
```

```python
df = pd.DataFrame({
    "name":["A","C","D","B","E","A","E","C"],
    "dept_id":["s1","s4","s5","s2","s3","s1","s4","s5"],
    "d_name" :["Analyst","Developer","Engineer","Trainer","UI Developer","Developer","Engineer","Analyst"]
}) 
```

```python
df
```

```python
x = df.drop(columns='d_name')
y = df['d_name']
```

```python
xtrain,xtest,ytrain,ytest = train_test_split(x,y,train_size=0.8,random_state=42)
```

```python
encoder = OneHotEncoder()
encoder.fit_transform(xtrain)
# it return Sparse Matrix
#o/p --> encoder = OneHotEncoder() encoder.fit_transform(xtrain)
# it return Sparse Matrix
```

```python
encoder = OneHotEncoder(sparse_output=False,handle_unknown='ignore')
encoder.fit_transform(xtrain)
```
